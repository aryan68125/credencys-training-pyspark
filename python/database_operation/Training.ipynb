```python
# default behaviour of zip
names=["Rollex","Ballistics","Barricade","Barbatos","Bullet","Kombat","ShockWave","Ignition","Dynamite","Shotgun","Artillery"]
roll=[997,1000,247,117,762,365,360,6000,911]

zip_list = list(zip(names,roll))
dict_result = {key:val for key,val in zip_list}
dict_result
```

```python
# Dictionary comprehension when elements are not same
import itertools

names=["Rollex","Ballistics","Barricade","Barbatos","Bullet","Kombat","ShockWave","Ignition","Dynamite","Shotgun","Artillery"]
roll=[997,1000,247,117,762,365,360,6000,911]

zip_list = list(zip(names,itertools.cycle(roll)))
dict_result = {key:val for key,val in zip_list}
dict_result
```

```python
# dictionary comprehension with conditional statements
names=["Rollex","Ballistics","Barricade","Barbatos","Bullet","Kombat","ShockWave","Ignition","Dynamite","Shotgun","Artillery"]
roll=[997,1000,247,117,762,365,360,6000,911,12,150]

zip_list = list(zip(names,roll))
dict_result = {key:val for key,val in zip_list if val%2==0}
dict_result
```

```python
numbers = range(10,70,10)
values = [2,5]

calc = []
for j in range(len(values)):
    for i in range(len(numbers)):
        print(f"{numbers[i]} * {values[j]} = {numbers[i]*values[j]}")
        calc.append(numbers[i]*values[j])
calc
```

```python
import itertools 
numbers = range(10,70,10)
numbers
values = [2,5]
values
zip_num = list(zip(numbers,itertools.cycle(values)))
for i,j in zip_num:
    print(f"{i} * {j} = {i*j}")
```
